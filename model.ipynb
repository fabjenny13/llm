```python
import torch
import tiktoken
from package import GPTModel, GPT_CONFIG_124M
```

```python
model = GPTModel(GPT_CONFIG_124M)
```

Cross Entropy

```python
input_text = ["Every effort moves", "Hello my name"]
target_text = [" effort moves you", " my name is"]
tokenizer = tiktoken.get_encoding("gpt2")


input_tokens = torch.tensor([tokenizer.encode(text) for  text in input_text])
target_tokens = torch.tensor([tokenizer.encode(text) for text in target_text])

print("Input: ", input_tokens)
print("Target: ", target_tokens)
```

```python
logits = model(input_tokens)
torch.nn.functional.cross_entropy(logits.flatten(0, 1), target_tokens.flatten())
```
